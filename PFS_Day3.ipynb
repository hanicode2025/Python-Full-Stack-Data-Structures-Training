```python
a = input()
vowels = []
consonants = []
for i in a:
    if i == "a" or i == "e" or i == "i" or i == "o" or i == "u" or i == "A" or i == "E" or i == "I" or i == "O" or i == "U":
        if i not in vowels:
            vowels.append(i)
    else:
        if i not in consonants:
            consonants.append(i)
print("Vowels ", vowels)
print("Consonants ", consonants)
```

```python
num = int(input("Enter number: "))
order = len(str(num))
total = 0
temp = num

while temp > 0:
    total += (temp % 10) ** order
    temp //= 10

if num == total:
    print("Armstrong number")
else:
    print("Not an armstrong number")
```

```python
#prime numbers in a range

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
for i in range(num1, num2+1):
    for j in range(2, i):
        if i % j == 0:
            break
    else:
        print(i)
```

```python
#simple calculator

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
op = input("Enter operator: ")
if op == "+":
    print(num1 + num2)
elif op == "-":
    print(num1 - num2)
elif op == "*":
    print(num1 * num2)
elif op == "/":
    print(num1 / num2)
else:
    print("Invalid operator")\
```

```python
#perfect number
per = int(input("Enter number: "))
sum = 0
for i in range(1, per):
    if per % i == 0:
        sum += i
if sum == per:
    print("Perfect number")
else:
    print("Not a perfect number")
```

```python
#using while loop count no of even digits in a number

num = int(input("Enter number: "))
count = 0
while num > 0:
    if num % 2 == 0:
        count += 1
    num //= 10
print(count)
```

```python
#longest word in the sentence

sentence = input("Enter sentence: ")
words = sentence.split()
longest = ""
for word in words:
    if len(word) > len(longest):
        longest = word
print(longest)
```

```python
# Simple solution using max()
sentence = input("Enter sentence: ")
words = sentence.split()

longest = max(words, key=len)
print(longest)
```

```python
# Fixed version using the reversed function
sentence = input("Enter Sentence: ")
words = sentence.split()
for word in reversed(words):
    print(word, end=" ")
```
