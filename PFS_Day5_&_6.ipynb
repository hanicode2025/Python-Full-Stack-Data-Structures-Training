```python
def BinarySearch(a,high,low,target):
  while low <= high:
    mid = (high + low) // 2
    if a[mid] == target:
      return mid
    elif a[mid] < target:
      low = mid + 1
    else:
      high = mid - 1
      return - 1
a = list(map(int, input().split()))
target = int(input())
print(BinarySearch(a,len(a)-1,0,target))
```

```python
def factorsofn(n):
    factors = []
    for i in range(1, n):
        if n % i == 0:
            factors.append(i)
    return factors

def isprime(n):
    if n > 1:
        for j in range(2, n):
            if n % j == 0:
                return False
        return True
    return False

m = int(input())
y = factorsofn(m)

prime_factors = [x for x in y if isprime(x)]
print(prime_factors)
```

```python
class Solution(object):
    def search(self, nums, target):
      left = 0
      right = len(nums) - 1
      while left <= right:
        mid = (left + right)//2
        if nums[mid] == target:
          return mid
        if nums[left] <= nums[mid]:
          if nums[left] <= target < nums[mid]:
            right = mid - 1
          else:
            left = mid + 1
        else:
          if nums[mid] < target <= nums[right]:
            left = mid + 1
          else:
            right = mid - 1
      return -1
```

```python
def alphabet(n):
  vowels = []
  consonants = []
  for i in n:
    if i.lower() in "aeiou" or i.upper() in "AEIOU":
      vowels.append(i)
    else:
      consonants.append(i)
  return vowels, consonants

m = input()
vowels, consonants = alphabet(m)
print("Vowels: ", vowels)
print("Consonants: ", consonants)
```

```python
b = list(map(int, input().split()))
a = [0] + b + [0]

peaks = []

for i in range(1, len(a) - 1):
    if a[i] > a[i - 1] and a[i] > a[i + 1]:
        peaks.append(i - 1)

if len(peaks) == 0:
    print("No peaks")
else:
    largest = peaks[0]
    for i in peaks:
        if i > largest:
            largest = i
    print(largest)
```

```python
class Solution(object):
  def findPeakElement(self,nums):
    left = 0
    right = len(nums) - 1
    while left < right:
      mid = (left + right) // 2
      if nums[mid] > nums[mid + 1]:
        left = mid + 1
      else:
        right = mid
    return left
```
